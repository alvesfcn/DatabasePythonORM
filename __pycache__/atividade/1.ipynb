```python
def numero(n1):
    return n1 * 2  

n1 = int(input("Digite um número: "))

print("O dobro do número", n1, "é", numero(n1))
```

```python
def numero(n1,n2):
  for a in range (n1,n2):
    print(a)
n1 = int(input("Digite um número: "))
n2 = int(input("Digite outro número: ")) 
numero(n1,n2)
  
```

```python
def numero(n1):
  if n1 % 2 == 0:
    print("O número", n1, "é par")
  else:
    print("O número",n1, "é ímpar")

n1 = int(input("Digite um número: "))
numero(n1)
```

```python
def numero(n1, n2, n3):
    return n1 + n2 + n3

n1 = int(input("Digite um número: "))
n2 = int(input("Digite outro número: "))
n3 = int(input("Digite mais um número: "))

soma = numero(n1, n2, n3)
media = soma / 3

print("A soma dos números é:", soma)
print("A média dos números é:", media)

```
